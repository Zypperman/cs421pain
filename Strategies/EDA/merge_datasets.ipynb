# Merge Datasets
 
Merge `first_batch_with_labels.npz` and `training_batch_with_labels.npz` into a single CSV.

```python
import numpy as np
import pandas as pd

# Load both datasets
f1 = np.load('../../ORIGINAL MATS DONT MODIFY/training_batch_with_labels.npz')
f2 = np.load('../../ORIGINAL MATS DONT MODIFY/first_batch_with_labels.npz')
f3 = np.load('../../ORIGINAL MATS DONT MODIFY/second_batch_with_labels.npz')

print("training_batch_with_labels  - X:", f1['X'].shape, "y:", f1['y'].shape)
print("first_batch_with_labels - X:", f2['X'].shape, "y:", f2['y'].shape)
print("second_batch_with_labels - X:", f3['X'].shape, "y:", f3['y'].shape)
```

```python
# Concatenate X and y arrays from all batches
X_merged = np.concatenate([f1['X'], f2['X'], f3['X']], axis=0)
y_merged = np.concatenate([f1['y'], f2['y'], f3['y']], axis=0)

print("Merged X:", X_merged.shape)
print("Merged y:", y_merged.shape)
```

```python
# Create DataFrames
# X has 3 columns: [id, feature1, feature2]
# y has 2 columns: [id, label]
df_X = pd.DataFrame(X_merged, columns=['id', 'feature1', 'feature2'])
df_y = pd.DataFrame(y_merged, columns=['id', 'label'])

# Merge X and y on id
df_merged = df_X.merge(df_y, on='id', how='left')

print(f"Merged DataFrame shape: {df_merged.shape}")
print(f"Unique IDs: {df_merged['id'].nunique()}")
df_merged.head(10)
```

```python
# Export to npz

np.savez('Labelled_0_1_2.npz', X=X_merged, y=y_merged)
print("Saved Labelled_0_1_2.npz")
```
